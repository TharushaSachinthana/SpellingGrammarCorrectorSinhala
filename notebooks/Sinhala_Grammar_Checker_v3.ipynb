```python
def grammar_checker_rule_based(sentence):
    corrections = []

    # Rule 1: If the sentence has "අපි" as the subject, the verb should end with "මු"
    if "අපි" in sentence:
        words = sentence.split()
        if words[-1].endswith("මු"):
            return ["The sentence is grammatically correct."]
        else:
            corrections.append(
                f"Incorrect verb ending for subject 'අපි'. Replace '{words[-1]}' with a verb ending in 'මු'."
            )

    # Rule 2: If the sentence has "මම" as the subject, the verb should end with "මි"
    if "මම" in sentence:
        words = sentence.split()
        if words[-1].endswith("මි"):
            return ["The sentence is grammatically correct."]
        else:
            corrections.append(
                f"Incorrect verb ending for subject 'මම'. Replace '{words[-1]}' with a verb ending in 'මි'."
            )

    # Rule 3: If sentence does not have either "මම" or "අපි" sentence ends with "යි"
    if "මම" not in sentence and "අපි" not in sentence:
        words = sentence.split()
        subject = words[0]
        for word in words:
            if word.endswith("මු"):
                corrections.append(
                    f"Incorrect verb ending for subject {subject}. Replace '{word}' with a verb ending in 'යි'."
                )
            if word.endswith("මි"):
                corrections.append(
                    f"Incorrect verb ending for subject {subject}. Replace '{word}' with a verb ending in 'යි'."
                )

    # Return the corrections or confirm correctness
    if corrections:
        return corrections
    else:
        return ["The sentence is grammatically correct."]

for test in test_sentences:
    result = grammar_checker_rule_based(test)
    print(f"Sentence: {test}")
    for res in result:
        print(f"  - {res}")
    print()

def calculate_accuracy_rule_based(test_sentences, correctness):
    correct_predictions = 0
    total_sentences = len(test_sentences)

    for sentence, true_label in zip(test_sentences, correctness):
        result = grammar_checker_rule_based(sentence)
        predicted_label = 1 if "grammatically correct" in result[0] else 0
        if predicted_label == true_label:
            correct_predictions += 1

    accuracy = correct_predictions / total_sentences
    return accuracy

rule_based_accuracy = calculate_accuracy_rule_based(test_sentences, correctness)
print(f"Rule-Based Method Accuracy: {rule_based_accuracy}")
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

# Dataset
sentences = [
    "මම ගීතයක් ගයමි",
    "මම බුඳුන් වදිමි",
    "අපි චාරිකාවක් යමු",
    "අපි ආහාර පිසිමු",
    "මම පොතක් කියවමි",
    "අපි පොත් කියවමු",
    "මම තේ කෝප්පයක් බොමි",
    "අපි නගරයට ගියමු",
    "මම කෑම පිසමි",
    "අපි දිනපොත් මිලදී ගමු",
    "මම ගමනක් යමි",
    "අපි ගමනක් යමු",
    "මම ගහක් යට විවේක ගනිමි",
    "අපි ගසක් යට විවේක ගනිමු",
    "මම කෑම කමි",
    "අපි ගසකට යමු"
]

# Labels (verb endings)
labels = ["මි", "මි", "මු", "මු", "මි", "මු", "මි", "මු", "මි", "මු", "මි", "මු", "මි", "මු", "මි", "මු"]

# Feature Engineering
def sentence_to_features(sentence):
    features = []
    words = sentence.split()

    # Subject Features
    features.append(1 if "මම" in words else 0)  # "මම" presence
    features.append(1 if "අපි" in words else 0)  # "අපි" presence

    # Verb Ending Features
    verb = words[-1] if words else ""
    features.append(1 if verb.endswith("මි") else 0)  # Ends with "මි"
    features.append(1 if verb.endswith("මු") else 0)  # Ends with "මු"

    return features

# Convert sentences to features
X = np.array([sentence_to_features(sentence) for sentence in sentences])

# Convert labels to numeric values
label_mapping = {"මි": 0, "මු": 1}
y = np.array([label_mapping[label] for label in labels])

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Model Training
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Predictions and Evaluation
y_pred = model.predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred))

# Prediction Function
def predict_subject_and_verb(sentence):
    features = np.array(sentence_to_features(sentence)).reshape(1, -1)
    predicted_label = model.predict(features)[0]
    subject = "මම" if features[0][0] == 1 else "අපි" if features[0][1] == 1 else "Unknown"
    verb_ending = [key for key, value in label_mapping.items() if value == predicted_label][0]
    return subject, verb_ending

# Predicted labels and grammar check
predicted_labels = []
actual_labels = []
predicted_correctness = []

for sentence in test_sentences:
    # Extract the actual verb ending by splitting the sentence
    actual_verb_ending = sentence.split()[-1][-2:]  # Last two characters as verb ending
    actual_labels.append(actual_verb_ending)

    # Predict the subject and verb ending
    subject, predicted_verb_ending = predict_subject_and_verb(sentence)
    predicted_labels.append(predicted_verb_ending)

    # Check correctness
    if predicted_verb_ending == actual_verb_ending:
        print(f"Sentence: {sentence} is grammatically correct.\n")
        predicted_correctness.append(1)
    else:
        print(f"Sentence: {sentence} is grammatically incorrect.\nPredicted Subject: {subject}, Predicted Verb Ending: {predicted_verb_ending}\n")
        predicted_correctness.append(0)

# Calculate accuracy
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(correctness, predicted_correctness)

print(f"ML Accuracy: {accuracy:.2f}")
```
